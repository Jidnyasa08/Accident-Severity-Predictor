```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
import joblib
```

```python
df= pd.read_csv(r"C:\Users\jidny\OneDrive\Documents\Hiresnix_Internship\Project_1_Accident_Severity_Predictor\Dataset\Road.csv")
```

```python
df.columns
```

```python
X = df.drop('Accident_severity' , axis=1)
Y = df['Accident_severity']
```

```python
X.select_dtypes(include=['object']).columns
```

```python
encoders = {}

for col in X.select_dtypes(include=['object']).columns:

    le = LabelEncoder()

    X[col] = le.fit_transform(X[col].astype(str))

    encoders[col] = le
```

```python
joblib.dump(
    encoders,
    r"C:\Users\jidny\OneDrive\Documents\Hiresnix_Internship\Project_1_Accident_Severity_Predictor\Model\encoders.pkl"
)
```

```python
target_encoder = LabelEncoder()
target_encoder = LabelEncoder()
Y = target_encoder.fit_transform(Y)
```
